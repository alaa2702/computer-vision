```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
image = cv2.imread('download.jpg')


choice = input("Choose transformation: darken, lighten, invert: ").strip().lower()

# Apply transformation
if choice == "darken":
    transformed = cv2.convertScaleAbs(image, alpha=0.5, beta=0)  # Reduce brightness
elif choice == "lighten":
    transformed = cv2.convertScaleAbs(image, alpha=1.5, beta=0)  # Increase brightness
elif choice == "invert":
    transformed = cv2.bitwise_not(image)  # Invert colors
else:
    print("Invalid choice.")
    exit()

# Show original and transformed images
cv2_imshow(image)
cv2_imshow(transformed)

```

```python

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

lower, upper = 100, 150

for i in range(gray.shape[0]):
    for j in range(gray.shape[1]):
        if lower <= gray[i, j] <= upper:
            gray[i, j] = 255
cv2_imshow(image)
cv2_imshow(gray)
```

```python
kernel = np.array([[0, -1, 0],
                   [-1, 5, -1],
                   [0, -1, 0]])

filtered = cv2.filter2D(image, -1, kernel)

cv2_imshow(image)
cv2_imshow(filtered)
```

```python
img=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
blur1 = cv2.GaussianBlur(img, (5, 5), 1)
blur2 = cv2.GaussianBlur(img, (5, 5), 2)


dog = cv2.absdiff(blur1, blur2)
dog_normalized = cv2.normalize(dog, None, 0, 255, cv2.NORM_MINMAX)

cv2_imshow(image)
cv2_imshow(dog_normalized)
```
